```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd()

# If the notebook is opened from the notebooks folder,
# move up one level to the project root.
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "transactions.csv"
OUTPUT_DIR = PROJECT_ROOT / "notebooks" / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project folder:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())
```

```python
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at: {DATA_PATH}\n"
        "Place your CSV in data/raw/ and name it transactions.csv."
    )

print(f"File size: {DATA_PATH.stat().st_size / (1024**2):.2f} MB")

preview = pd.read_csv(DATA_PATH, nrows=10)

display(preview)

print("Columns:")
print(preview.columns.tolist())

print("\nData types:")
display(preview.dtypes)
```

```python
df = pd.read_csv(DATA_PATH, low_memory=False)

print("Dataset shape:", df.shape)

print("\nData types:")
display(df.dtypes)

print("\nStatistical summary:")
display(df.describe(include="all").T)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())

if "isFraud" in df.columns:
    print("\nFraud label counts:")
    display(df["isFraud"].value_counts(dropna=False))

    print("\nFraud label percentages:")
    display(
        (df["isFraud"].value_counts(normalize=True, dropna=False) * 100)
        .rename("percentage")
    )

print(
    "\nMemory used by DataFrame:",
    f"{df.memory_usage(deep=True).sum() / (1024**2):.2f} MB"
)
```

```python
df.head(100).to_csv(
    OUTPUT_DIR / "dataset_preview.csv",
    index=False
)

df.dtypes.astype(str).rename("dtype").to_csv(
    OUTPUT_DIR / "column_types.csv"
)

df.isnull().sum().rename("missing_count").to_csv(
    OUTPUT_DIR / "missing_values.csv"
)

print("Inspection reports saved to:", OUTPUT_DIR)
```
